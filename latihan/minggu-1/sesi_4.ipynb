```python
import numpy as np

penjualan = np.array([
    [12, 8, 5, 10],
    [15, 9, 7, 11],
    [10, 6, 4, 9]])

# Rata-rata dan standar deviasi tiap menu
mu = penjualan.mean(axis=0)
sigma = penjualan.std(axis=0)
print(mu.shape, sigma.shape)

# Z-score per menu
z_menu = (penjualan - mu) / sigma
print(z_menu.round(2))
```

```python
mu_h = penjualan.mean(axis=1, keepdims=True)
sigma_h = penjualan.std(axis=1, keepdims=True)
print(mu_h.shape)

z_hari = (penjualan - mu_h) / sigma_h
print(z_hari.round(2))
```

#### Note
- standar deviasi (σ): ukuran seberapa jauh nilai-nilai menyebar dari rata-ratanya. Ini untuk mengukur area penyebaran data
    
    __Langkah menghitung nilai standar deviasi:__
    1. hitung rata-rata populasi atau kumpulan data atau nilai.
    2. kurangkan semua nilai atau data dengan rata rata
    3. kuadratkan semua nilai hasil pengurangan tadi
    4. jumlahkan seluruh nilai nya
    5. nilai hasil penjumlahan di bagi dengan total banyaknya data atau nilai
    6. terakhir, akarkan hasil pembagian tadi (inilah hasil atau nilai standar deviasi)

    __Cara menentukan area sebaran data (area banyaknya data berkumpul):__
    1. ambil nilai rata-rata (contoh: 70) dan nilai standar deviasi (contoh: 10)
    2. tentukan batas atas (rata-rata + standar deviasi) --> 70 + 10 = 80
    3. tentukan batas bawah (rata-rata - standar deviasi) --> 70 - 10 = 60
    4. area = batas bawah sampai batas atas (60 - 80), jadi area tempat data banyak berkumpul berada diantara 60 sampai 80

- rata-rata (μ): jumlah semua nilai dibagi banyaknya nilai.

- Z-score: jarak sebuah titik nilai dari rata-rata, dihitung dalam satuan standar deviasi.

    __Langakah menghitung z-score:__
    1. ambil rata-rata (contoh: 70), nilai standar deviasi (contoh: 10), dan nilai tujuan atau yang ingin di cek (contoh: 90)
    2. rumusnya: (nilai tujuan - rata-rata) / standar deviasi, (90 - 70) / 10 = 2 --> z-score untuk nilai atau data 90 adalah 2
    3. hasil z-score positif = diatas rata-rata sedangkan negatif = dibawah rata-rata
    3. cara baca data: jarak antara nilai 90 dengan rata rata adalah 2 z-score atau 2 kali standar deviasi diatas rata-rata

- parameter __keepdims__ berfungsi untuk mempertahankan dimensi sebuah array saat melakukan agregasi

    contoh: Menghitung rata-rata dengan axis=1
    
    [[2, 2],

    [3, 3]]

    jika kita hitung array 2d diatas pakai .mean denagn axis=1 maka ukurannya akan berubah jadi array 1 dimensi (2,). Dengan keepdims apakan mempertahankan dimensinya sehingga hasilnya (2, 1) --> tetap array 2 dimensi
