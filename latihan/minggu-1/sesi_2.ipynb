```python
import numpy as np

penjualan = np.array([
    [10, 20, 30, 40],
    [5, 15, 25, 35],
    [1, 2, 3, 4],
])

kenaikan_menu = np.array([1,2,3,4])

print(kenaikan_menu.shape)
print(penjualan + 2)
print(penjualan + kenaikan_menu)
```

```python
kenaikan_hari = np.array([100,200,300]).reshape(3,1)

print(kenaikan_hari.shape)
print(penjualan + kenaikan_hari)
```

## Aturan broadcasting

Broadcasting adalah cara NumPy mengoperasikan dua array yang shape-nya beda, dengan "membagikan" array yang lebih kecil ke array yang lebih besar tanpa menyalinnya.

### Cara mengeceknya

1. Tulis kedua shape sejajar, **rata kanan**.
2. Bandingkan tiap pasangan angka yang berhadapan, dari **kanan ke kiri**.
3. Sepasang angka **cocok** kalau:
   - keduanya **sama**, atau
   - salah satunya bernilai **1** (angka 1 dibagikan ke angka pasangannya).
4. Angka yang **tidak punya pasangan** (karena shape-nya lebih pendek) dibiarkan lolos.
5. Kalau ada **satu saja** pasangan yang tidak cocok, hasilnya **error**.

Shape hasilnya: untuk tiap pasangan, ambil angka yang lebih besar.

### Contoh dari praktik sendiri

| Operasi | Hasil | Alasan |
|---|---|---|
| `(3, 4)` + `(4,)` | berhasil, `(3, 4)` | 4 dengan 4 sama; angka 3 tidak punya pasangan |
| `(3, 4)` + `(3,)` | error | 4 dengan 3 tidak sama dan tidak ada yang 1 |
| `(3, 4)` + `(3, 1)` | berhasil, `(3, 4)` | 4 dengan 1 lolos; 3 dengan 3 sama |
| `(3, 4)` + `(2, 4)` | error | 3 dengan 2 tidak sama dan tidak ada yang 1 |
| `(3, 1)` + `(1, 4)` | berhasil, `(3, 4)` | 1 dengan 4 lolos; 3 dengan 1 lolos |
| `(5, 1)` + `(1, 6)` | berhasil, `(5, 6)` | kedua pasangan lolos karena ada angka 1 |

### Hal yang perlu diingat

- Array 1 dimensi selalu dicocokkan dengan **kolom** (sisi kanan), makanya angkanya terbagi ke bawah.
- Untuk membagikan angka **ke samping** (misalnya kenaikan per hari), ubah shape jadi `(n, 1)` dengan `reshape(n, 1)`.
- Yang dicek adalah **posisi berhadapan dari kanan**, bukan sekadar "ada angka yang sama" di shape lain.
