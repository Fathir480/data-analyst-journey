```python
import numpy as np

penjualan = np.array([
    [12, 8, 5, 10],
    [15, 9, 7, 11],
    [10, 6, 4, 9]])
harga = np.array([18000, 22000, 25000, 15000])

pendapatan_loop = np.zeros((3, 4))
for i in range(3):
    for j in range(4):
        pendapatan_loop[i, j] = penjualan[i, j] * harga[j]

pendapatan = penjualan * harga
```

```python
print(pendapatan_loop.shape, pendapatan.shape, pendapatan_loop[1, 2])
print(np.array_equal(pendapatan_loop, pendapatan))
```
