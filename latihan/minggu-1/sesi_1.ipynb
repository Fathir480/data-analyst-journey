# Minggu 1, Sesi 1: NumPy, konsep axis

**Tujuan sesi:** paham axis sebagai arah di dalam data, dan bisa menebak shape hasil `sum`.

**Data:** penjualan warung kopi, 3 hari × 4 menu.

```python
import numpy as np

penjualan = np.array([
    [10, 20, 30, 40],
    [5, 15, 25, 35],
    [1, 2, 3, 4]
])

print(penjualan.shape)
print(penjualan.sum(axis=0))
print(penjualan.sum(axis=1))
print(penjualan.sum(axis=0).shape)
print(penjualan.sum(axis=1).shape)
```

Axis merupakan sebuah parameter atau keyword yang akan menunjukkan arah dari sebuah array 2 Dimensi.

Dari contoh diatas, terdapat data penjualan yang memiliki 2 arah (baris, kolom), 1 arah horizontal (baris) yang menunjukkan arah hari, dan arah lainnya vertikal (kolom) yang menunjukkan arah menu.

Dalam kode .sum(axis=) ini artinya menjumlahkan dengan menghilangkan salah satu arah dengan cara melakukan penjumlahan berdasarkan arah tersebut. Misal untuk axis=0 yang artinya menghilangkan arah verikal dengan menjumlahkan nilai dari semua baris untuk setiap kolom, sehingga hasil shape atau ukuran datanya adalah arah yang tersisa yaitu 4 atau sama dengan jumlah kolom.

### Note
- Axis adalah penunjuk arah di dalam data. 
    - axis=0 menurun (antar baris) # Menjumlahkan semua baris untuk setiap kolom
    - axis=1 menyamping (antar kolom) # Menjumlahkan semua kolom untuk setiap baris
